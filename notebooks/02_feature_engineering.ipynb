This notebook demonstrates feature importance ranking, robust telemetry scaling, and clean data partitioning.

```python
import os
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Add project root
sys.path.append(os.path.abspath('..'))
from src.data_loader import load_dataset
from src.preprocessor import NIDSPreprocessor

sns.set_theme(style="whitegrid")
%matplotlib inline
```

### 1. Ingest Raw Dataset & Fit Preprocessor

```python
df = load_dataset(sample_size=30000)
preprocessor = NIDSPreprocessor(top_k_features=25)
processed_df = preprocessor.fit_transform(df, label_col="Label")
print(f"Processed Data Shape: {processed_df.shape}")
processed_df.head()
```

### 2. Visualize Feature Importance Rankings

```python
plt.figure(figsize=(10, 6))
importance_series = pd.Series(preprocessor.feature_importance_scores)
importance_series.sort_values(ascending=True).plot(kind='barh', color='teal')
plt.title('Top 25 Selected Network Telemetry Features (Importance Score)')
plt.xlabel('Gini Importance Score')
plt.ylabel('Flow Feature')
plt.tight_layout()
plt.show()
```

### 3. Scaling Distribution Inspection

```python
fig, ax = plt.subplots(figsize=(12, 5))
features_to_plot = preprocessor.selected_features[:6]
sns.boxplot(data=processed_df[features_to_plot], ax=ax, palette="Set2")
ax.set_title("RobustScaled Telemetry Distribution (Top 6 Features)")
ax.set_xticklabels(features_to_plot, rotation=30, ha="right")
plt.show()
```
